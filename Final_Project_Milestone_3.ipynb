# Loading the csv file

```python
import pandas as pd


df = pd.read_csv("sport car price.csv")
df.head

```

# Below i"m dropping the rows that have no values 

```python

df = df.dropna(subset=["Engine Size (L)", "Torque (lb-ft)"])
```

# A lot of these columns come with data that is a number and a word in the same entry like "500lb-ft" or "200 hp ". So i had to clean them so they're usuable and not hindering the process. Now they will just be numbers like 200 or 700 !

# I didn't know how to do this from memory so I searched up on Google and it opened panda py data . But that looked too complex. So I had to look it up on different websites and then I got it. It was a mixture of Reddit and YouTube.

```python
# Remove any non-numeric characters using pandas string methods
df["Horsepower"] = df["Horsepower"].str.replace(" hp", "", regex=False)
df["Horsepower"] = pd.to_numeric(df["Horsepower"], errors="coerce")

df["Torque (lb-ft)"] = df["Torque (lb-ft)"].str.replace(" lb-ft", "", regex=False)
df["Torque (lb-ft)"] = pd.to_numeric(df["Torque (lb-ft)"], errors="coerce")

df["0-60 MPH Time (seconds)"] = df["0-60 MPH Time (seconds)"].str.replace(" sec", "", regex=False)
df["0-60 MPH Time (seconds)"] = pd.to_numeric(df["0-60 MPH Time (seconds)"], errors="coerce")

df["Price (in USD)"] = df["Price (in USD)"].str.replace(",", "").str.replace("$", "", regex=False)
df["Price (in USD)"] = pd.to_numeric(df["Price (in USD)"], errors="coerce")
```

# for all the roles that didn't convert well or just left with a NA I had to remove them sadly 

```python
# Drop rows that failed conversion
df = df.dropna(subset=["Horsepower", "Torque (lb-ft)", "0-60 MPH Time (seconds)", "Price (in USD)"])
```

# I then renamed all the columns, cause it was not easy to work with 

```python
# Rename columns for easier use
df = df.rename(columns={
    "Car Make": "car_make",
    "Car Model": "car_model",
    "Year": "year",
    "Engine Size (L)": "engine_size_l",
    "Horsepower": "horsepower",
    "Torque (lb-ft)": "torque_lb_ft",
    "0-60 MPH Time (seconds)": "zero_to_60mph_s",
    "Price (in USD)": "price_usd"
})
```

# app.py

import pandas as pd
from dash import Dash, dcc, html, Input, Output
import plotly.express as px



# Ensure correct datatypes
df["engine_size_l"] = pd.to_numeric(df["engine_size_l"], errors="coerce")
df["horsepower"] = pd.to_numeric(df["horsepower"], errors="coerce")
df["torque_lb_ft"] = pd.to_numeric(df["torque_lb_ft"], errors="coerce")
df["zero_to_60mph_s"] = pd.to_numeric(df["zero_to_60mph_s"], errors="coerce")
df["price_usd"] = pd.to_numeric(df["price_usd"], errors="coerce")
df["year"] = pd.to_numeric(df["year"], errors="coerce")

# Initialize the app
app = Dash(__name__)
app.title = "🏎️ Sports Car Dashboard"

# Layout
app.layout = html.Div([
    html.H1("🏎️ Sports Car Performance and Price Explorer", style={'textAlign': 'center'}),

    # Filters
    html.Div([
        html.Label("Select Make:"),
        dcc.Dropdown(
            id="make-dropdown",
            options=[{"label": make, "value": make} for make in df["car_make"].unique()],
            multi=True,
            value=df["car_make"].unique()[:3]  # default selection
        )
    ], style={'width': '50%', 'padding': '20px'}),

    # Graphs
    html.Div([
        dcc.Graph(id="scatter-hp-price"),
        dcc.Graph(id="bar-price-trend"),
    ])
])

# Callback to update graphs
@app.callback(
    [
        Output("scatter-hp-price", "figure"),
        Output("bar-price-trend", "figure")
    ],
    [
        Input("make-dropdown", "value")
    ]
)
def update_graphs(selected_makes):
    if not selected_makes:
        empty_fig = px.scatter(title="No Data Available")
        return empty_fig, empty_fig, empty_fig, empty_fig

    # Filter the data
    dff = df[df["car_make"].isin(selected_makes)]

    if dff.empty:
        empty_fig = px.scatter(title="No Cars Found")
        return empty_fig, empty_fig, empty_fig, empty_fig

    # Scatter Plot
    fig_scatter = px.scatter(
        dff,
        x="horsepower",
        y="price_usd",
        color="car_make",
        size="engine_size_l",
        hover_data=["car_model", "year"],
        title="Horsepower vs Price"
    )


    # Bar Chart
    avg_price_make = dff.groupby("car_make")["price_usd"].mean().reset_index()
    fig_bar = px.bar(
        avg_price_make,
        x="car_make",
        y="price_usd",
        text_auto=True,
        title="Average Price by Car Brand"
    )
    fig_bar.update_layout(xaxis={'categoryorder': 'total descending'})

    return fig_scatter, fig_bar

# Run app
if __name__ == '__main__':
    app.run_server(debug=True)
