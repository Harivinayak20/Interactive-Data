```python
import pandas as pd
import plotly.graph_objects as go
```

```python
meteor = pd.read_csv('Meteorite_Landings_Sample.csv')
meteor.head()
```

```python
meteor.index
```

```python
fig1 = go.Figure(data=go.Scattergeo(
    lon = meteor['reclong'],
    lat = meteor['reclat'],
    mode = 'markers',
))

fig1.update_layout(
    title = 'Meteorite Landings' ,
    geo_scope = 'world' ,
)

fig1.show()
```

```python
fig2 = go.Figure(data=go.Scattergeo(
    lon = meteor['reclong'],
    lat = meteor['reclat'],
    mode = 'markers',
    marker = dict(
        size = 8,
        opacity = 0.8,
        symbol = 'star-diamond',
        color = 'red'
    )
))

fig2.update_geos(projection_type = 'equirectangular')


fig2.update_layout(
    title = 'Meteorite Landings' ,
    geo_scope = 'world' ,
)

fig2.show()
```

```python
fig3 = go.Figure(data=go.Scattergeo(
    lon = meteor['reclong'],
    lat = meteor['reclat'],
    mode = 'markers',
    marker = dict(
        size = 8,
        opacity = 0.8,
        reversescale = True,
        autocolorscale = False,
        symbol = 'circle',
        colorscale = 'electric',
        color = meteor['mass (g)'],
        colorbar = dict(
            title = dict(
                text = 'Meteorite Masssss (g)'
            )
        )       
    )
))

fig3.update_geos(
    projection_type = 'equirectangular',
    showland = True , landcolor = 'lightgrey',
    showocean = True, oceancolor = 'light blue',
    showlakes = True, lakecolor = 'grey',
    showrivers = True , rivercolor = 'grey')


fig3.update_layout(
    title = 'Meteorite Landings' ,
    geo_scope = 'world' ,
)

fig3.show()
```

```python
fig4 = go.Figure(data=go.Scattergeo(
    lon = meteor['reclong'],
    lat = meteor['reclat'],
    mode = 'markers',
    marker = dict(
        size = 8,
        opacity = 0.8,
        reversescale = True,
        autocolorscale = False,
        symbol = 'circle',
        colorscale = 'electric',
        color = meteor['mass (g)'],
        colorbar = dict(
            title = dict(
                text = 'Meteorite Masssss (g)'
            )
        )       
    )
))

fig4.update_geos(
    projection_type = 'natural earth',
    showland = True , landcolor = 'light grey',
    showocean = True, oceancolor = 'light blue',
    showlakes = True, lakecolor = 'grey',
    showrivers = True , rivercolor = 'grey')


fig4.update_layout(
    title = 'Meteorite Landings' ,
    geo_scope = 'world' ,
)

fig4.show()
```

```python
fig5 = go.Figure()

fig5.add_trace(
        go.Scattergeo(
        lon = meteor['reclong'],
        lat = meteor['reclat'],
        mode = 'markers',
        marker = dict(
        size = 8,
        opacity = 0.8,
        reversescale = True,
        autocolorscale = False,
        symbol = 'circle',
        colorscale = 'electric',
        color = meteor['mass (g)'],
        colorbar = dict(
            title = dict(
                text = 'Meteorite Masssss (g)'
            )
        )       
    )
))


fig5.update_geos(
    projection_type = 'equirectangular',
    showland = True , landcolor = 'light grey',
    showocean = True, oceancolor = 'light blue',
    showlakes = True, lakecolor = 'grey',
    showrivers = True , rivercolor = 'grey')


button1 = {
    'args' : [{'geo' : {'projection' : {'type' : 'orthographic'}}}],
              'label' : 'map',
              'method' : 'relayout'
}


button2 = {
    'args' : [{'geo' : {'projection' : {'type' : 'equirectangular'}}}],
              'label' : 'Globe',
              'method' : 'relayout'
}

fig5.update_layout(
    title = 'Meteorite Landings' ,
    geo_scope = 'world' ,
    updatemenus = [
        dict(
            buttons = [button1, button2],
            direction = 'down',
            x = 0.1,
            xanchor = 'left',
            y = 1.1, 
            yanchor = 'top'
        )
    ]
)


fig5.show()
```

```python
meteor.columns = [col.replace("mass (g)", "mass") for col in meteor.columns]


fig6 = go.Figure()

fig6.add_trace(
    go.Bar(
        x = meteor.year, y = meteor.mass
    )
)
fig6.update_layout(
    title_text = "meteors' Mass by Year"
)

fig6.show()
```
