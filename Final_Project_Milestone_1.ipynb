# Loading the csv file

```python
import pandas as pd


df = pd.read_csv("sport car price.csv")


```

# Below i"m dropping the rows that have no values 

```python

df = df.dropna(subset=["Engine Size (L)", "Torque (lb-ft)"])
```

# A lot of these columns come with data that is a number and a word in the same entry like "500lb-ft" or "200 hp ". So i had to clean them so they're usuable and not hindering the process. Now they will just be numbers like 200 or 700 !

# I didn't know how to do this from memory so I searched up on Google and it opened panda py data . But that looked too complex. So I had to look it up on different websites and then I got it. It was a mixture of Reddit and YouTube.

```python
# Remove any non-numeric characters using pandas string methods
df["Horsepower"] = df["Horsepower"].str.replace(" hp", "", regex=False)
df["Horsepower"] = pd.to_numeric(df["Horsepower"], errors="coerce")

df["Torque (lb-ft)"] = df["Torque (lb-ft)"].str.replace(" lb-ft", "", regex=False)
df["Torque (lb-ft)"] = pd.to_numeric(df["Torque (lb-ft)"], errors="coerce")

df["0-60 MPH Time (seconds)"] = df["0-60 MPH Time (seconds)"].str.replace(" sec", "", regex=False)
df["0-60 MPH Time (seconds)"] = pd.to_numeric(df["0-60 MPH Time (seconds)"], errors="coerce")

df["Price (in USD)"] = df["Price (in USD)"].str.replace(",", "").str.replace("$", "", regex=False)
df["Price (in USD)"] = pd.to_numeric(df["Price (in USD)"], errors="coerce")
```

# for all the roles that didn't convert well or just left with a NA I had to remove them sadly 

```python
# Drop rows that failed conversion
df = df.dropna(subset=["Horsepower", "Torque (lb-ft)", "0-60 MPH Time (seconds)", "Price (in USD)"])
```

# I then renamed all the columns, cause it was not easy to work with 

```python
# Rename columns for easier use
df = df.rename(columns={
    "Car Make": "car_make",
    "Car Model": "car_model",
    "Year": "year",
    "Engine Size (L)": "engine_size_l",
    "Horsepower": "horsepower",
    "Torque (lb-ft)": "torque_lb_ft",
    "0-60 MPH Time (seconds)": "zero_to_60mph_s",
    "Price (in USD)": "price_usd"
})
```

# I saved all of the new progress into another CSV file because it's easier to distinguish and use for milestone two

```python
# cleaned csv file
df.to_csv("sport_car_price_cleaned.csv", index=False)


print("✅ Done. Preview of cleaned data:")
print(df.head())
```
